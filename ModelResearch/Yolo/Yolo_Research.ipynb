```python
from ultralytics import YOLO

dataset =  "dataset_path.yaml"
epochs = 5
img_size = 640
```

```python
model = YOLO("yolo11n-seg.pt") 
model.train(
    data=dataset,
    epochs=epochs,
    imgsz=img_size,
    project="runs/train",
    name="yolo11n",
    exist_ok=True
)
```
