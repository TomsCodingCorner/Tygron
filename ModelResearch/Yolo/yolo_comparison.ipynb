```python
from ultralytics import YOLO
import sys
from pathlib import Path
sys.path.append(str(Path().resolve().parents[1]))
from paths import YOLO_DATASET_YAML
```

```python
models = [
    "yolo11n-seg.pt",
    "yolo11s-seg.pt",
    "yolo11m-seg.pt",
    "yolo11l-seg.pt",
    "yolo11x-seg.pt",
    "yolov9e-seg.pt"
]

for model_path in models:
    print(f"\nEvaluating {model_path}")
    model = YOLO(model_path)

    results = model.val(data=YOLO_DATASET_YAML, split="val", save=False)

    map50 = results.seg.map50
    print(f"mAP@0.5 (Seg) for {model_path}: {map50:.4f}")
```

# MaP@0.5 (Mean average Precision met 0.5 confidence score) Resultaten
yolo11n-seg: 0.0074  
yolo11s-seg: 0.0046  
yolo11m-seg: 0.0044  
yolo11l-seg: 0.0059  
yolo11x-seg: 0.0045  
yolov9e-seg: 0.0037  

Opvallend is dat de prestaties van de modellen niet consistent verbeteren naarmate de modelgrootte toeneemt, wat doorgaans wel verwacht zou worden. Zo behaalt het kleinste model (yolo11n-seg) met 0.0074 de hoogste MaP@0.5-score, terwijl grotere modellen zoals yolo11x-seg en yolov9e-seg juist lagere scores laten zien.

Dit resultaat is opmerkelijk en suggereert dat schaalvergroting van het model in deze context niet automatisch leidt tot betere prestaties. Mogelijke verklaringen kunnen liggen in:
- Overfitting bij grotere modellen, vooral bij beperkte of suboptimale trainingsdata.
- Niet-geoptimaliseerde hyperparameters voor elk model.
- Mogelijke incompatibiliteit tussen de dataset en de capaciteiten van grotere modellen
